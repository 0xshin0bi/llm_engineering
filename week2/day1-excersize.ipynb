```python
import os
from dotenv import load_dotenv
from scraper import fetch_website_contents
from IPython.display import Markdown, display, update_display
from openai import OpenAI

OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')
OLLAMA_BASE_URL = "http://localhost:11434/v1"

system_prompt=""" you are a person in a conversation. in the user prompt, you have the name personality you need to respond as, as well as the names of the people in the conversation.YOU ARE TO ONLY RESPOND ONE TIME AS THE PERSON DESCRIBED IN THE USER PROMPT. 
the response should only be the answer. a good answer will start with the person's name. for example: ### JAKE: ... keep the answers up to one sentence."""

THE_CONVERSATION = "###AMIT: i hate working so much... also going to the gym is bad. i hate my life!\n"


amit_user_prompt = f""""your name is amit. you are a low self esteem whiner that like to complain. you are in a conversation with two of your friends, Blake and Jake. ANSWER ONLY AS AMIT.
this is the conversation so far: \n{THE_CONVERSATION}."""

jake_user_prompt = f""""your name is jake. you are an aggresive individual that likes to go to the gym and argue. you are in a conversation with two of your friends, Amit and Blake. ANSWER ONLY AS JAKE 
this is the conversation so far: \n{THE_CONVERSATION}. ."""

blake_user_prompt = f""""your name is blake. you are a african american human rights activist. you are in a conversation with two of your friends, Amit and Jake. ANSWER ONLY AS BLAKE. 
this is the conversation so far: \n{THE_CONVERSATION}.."""

openai = OpenAI()



def append_to_conversation(message):
    global THE_CONVERSATION
    THE_CONVERSATION += f"{message}\n"



def get_message_from_llama(user_prompt):
    stream = openai.chat.completions.create(model="gpt-5-nano", messages=[{"role": "system", "content": system_prompt, "role":"user","content":user_prompt}],stream=True)
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
    append_to_conversation(response)
    return response



display(Markdown(THE_CONVERSATION))

i=0
while i<5:
    #jake
    response = get_message_from_llama(jake_user_prompt)
    response = get_message_from_llama(blake_user_prompt)
    response = get_message_from_llama(amit_user_prompt)
    i+=1



```
